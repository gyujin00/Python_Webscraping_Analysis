```python
import pymysql  # MySQL 데이터베이스에 연결하기 위한 라이브러리

# MySQL DB와 연결 (호스트, 포트, DB명, 사용자, 비밀번호, 문자셋 설정)
db = pymysql.connect(
    host='localhost',
    port=3306,
    database='python_db',
    user='python',
    password='python',
    charset='utf8'
)

try:
    # cursor를 생성하고, with 문을 사용해 자동으로 close 처리
    with db.cursor() as cursor:
        
        # 기존 product 테이블을 삭제하는 SQL (필요 시 사용)
        # cursor.execute('drop table product')
        
        # product 테이블을 생성하는 SQL 실행
        cursor.execute(sql)
        
        # 테이블 생성 내용을 DB에 실제 반영
        db.commit()
        
        # 제품 데이터를 여러 건 삽입 (S2010 ~ S2019)
        for num in range(10, 20):  # 10부터 19까지 반복
            name = 'S20' + str(num)  # 제품명 생성 (예: S2010)
            
            # 파라미터 바인딩 방식의 INSERT SQL 작성
            ins_sql = \
                'insert into product (name, model_num, model_type) values (%s, %s, %s)'
            
            # SQL 실행 (값은 튜플로 전달하여 SQL 인젝션 방지)
            cursor.execute(ins_sql, (name, '7700', 'Phone'))
            
            # 실행된 SQL 문 출력 (확인용)
            print(ins_sql)

            # 아래는 문자열 결합 방식의 INSERT 예시 (보안상 권장되지 않음)
            # ins_sql = "insert into product (name,model_num,model_type) \
            # values('"+name+"','7700','Phone')"
            # cursor.execute(ins_sql)
        
        # 모든 INSERT 작업을 DB에 반영
        db.commit()
        
        # 마지막으로 삽입된 레코드의 ID 출력
        print(cursor.lastrowid)

except Exception as exp:
    # 예외 발생 시 오류 메시지 출력
    print(exp)
    
    # 오류가 발생하면 변경 사항을 모두 되돌림
    db.rollback()

finally:
    # DB 연결 종료
    db.close()
```

```python
import pymysql  # MySQL 데이터베이스와 연결하기 위한 라이브러리

# MySQL DB에 연결 (호스트, 포트, DB명, 사용자, 비밀번호, 문자셋 설정)
db = pymysql.connect(
    host='localhost',
    port=3306,
    database='python_db',
    user='python',
    password='python',
    charset='utf8'
)

try:
    # select, update 쿼리를 실행하기 위한 cursor 생성 (with 문으로 자동 close 처리)
    with db.cursor() as cursor:
        
        # id가 3인 상품 데이터를 조회
        cursor.execute('select * from product where id=3')
        
        # 조회 결과 중 한 행만 가져옴
        result = cursor.fetchone()
        
        # 결과 타입, 전체 row, 두 번째 컬럼 값 출력
        print(type(result), result, result[1])
        
        # 특정 이름 범위(S2010 ~ S2015)에 해당하는 상품의 model_type을 '핸드폰'으로 수정
        upd_sql = \
        "update product set model_type='%s' \
        where name between 'S2010' and 'S2015'" % '핸드폰'
        
        # update 쿼리 실행
        cursor.execute(upd_sql)
        
        # 변경 내용을 DB에 반영
        db.commit()
        
        # update로 영향을 받은 row 개수 출력
        print(cursor.rowcount)
        
        # product 테이블의 모든 데이터 조회
        cursor.execute('select * from product')
        
        # 조회 결과 전체를 리스트 형태로 가져옴
        result_list = cursor.fetchall()
        print(type(result_list))
        
        # 각 row의 컬럼 값 출력
        for row in result_list:
            print(row[0], row[1], row[2], row[3])
            
        # model_type 기준으로 그룹화하여 개수를 세는 쿼리 실행
        cursor.execute(
            'select model_type, count(*) from product group by model_type'
        )
        
        # 그룹화 결과 출력
        for row in cursor.fetchall():
            print(row)

finally:
    # DB 연결 종료
    db.close()
```

```python
# pymysql 라이브러리를 사용해 MySQL DB에 연결하기
import pymysql

# python_db 데이터베이스에 접속 정보 설정
con = pymysql.connect(
    host='localhost',
    port=3306,
    database='python_db',
    user='python',
    password='python',
    charset='utf8'
)

# print(type(con), con)  # 연결 객체 확인용 (디버깅 시 사용)

try:
    # cursor를 생성하고 with 문으로 자동 close 처리
    with con.cursor() as cursor:
        
        # name 컬럼 값이 지정된 값들에 해당하는 행을 삭제하는 SQL (IN 구문 사용)
        sql = "delete from product where name in (%s, %s)"
        
        # 삭제 쿼리 실행 (S2012, S2013에 해당하는 데이터 삭제)
        cursor.execute(sql, ('S2012', 'S2013'))
        
        # 삭제 내용을 DB에 실제 반영
        con.commit()
        
        # 삭제된 행(row)의 개수 출력
        print(cursor.rowcount)
                
        # product 테이블의 모든 데이터를 id 기준 오름차순으로 조회
        sql = "select * from product order by id"
        cursor.execute(sql)
        
        # 조회된 결과를 한 행씩 출력
        for row in cursor.fetchall():
            print(row[0], row[1], row[2], row[3])
            
except Exception as ex:
    # 예외 발생 시 변경 사항을 되돌림
    con.rollback()
    print(ex)

finally:
    # DB 연결 종료
    con.close()
```
