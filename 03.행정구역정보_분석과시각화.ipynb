```python
#광역시도명을 인자로 받아서 해당 광역시에 속한 행정구역들의 인구수의 평균과 편차 구하는 함수
def pop_mean_std(df, sido_name):
    print(f"{sido_name} 인구수의 평균과 편차")
    sido_df = df.loc[df['광역시도'] == sido_name, '인구수' : '행정구역'].copy() #'인구수' : '행정구역' =>인구수 ~ 행정구역 열 선택 
    print(f"{sido_name} 인구수 평균 = {sido_df['인구수'].mean()}")
    print(f"{sido_name} 인구수 편차 = {sido_df['인구수'].std()}")
```

### GroupBy 기능 사용하기
* 광역시도별 행정구역의 인구수 합계(서울시의 인구 합계, 경기도의 인구합계 등등)
* ~별에 해당하는 컬럼명이나 컬럼값을 groupby() 함수의 인자로 전달한다.
* Series 객체의 groupby(data['광역시도']) 함수는 "컬럼의 값"을 인자로 전달하고
* DataFrame 객체의 groupby('광역시도') 함수는 "컬럼명"을 인자로 전달합니다.
* 

```python
# Series 객체 사용(컬럼의 값을 인자로 전달) - 광역시도별 인구수의 합계 
# SeriesGroupBy object
data['인구수'].groupby(data['광역시도']).sum()
```

```python
# DataFrame 객체 사용(컬럼 명을 인자에 전달) - 광역시도별 인구수의 합계 
# DataFrameGroupBy object
# numeric_only = 숫자로 된 열만 알아서 골라서 계산해라.
data.groupby('광역시도').sum(numeric_only = True)
```

```python
# DataFrameGroupBy 
#groupby로 광역시도로 묶인 새 테이블이 생성되고, 그 테이블에 인구수만 뽑아서 sum함.
data.groupby('광역시도')['인구수'].sum()
```

```python
# 광역시도별 행정구역별 인구수의 합계
data.groupby(['광역시도','행정구역']).sum(numeric_only=True)
```

```python
# 광역시도와 행정구역별로 데이터를 그룹화한 뒤 각 지역의 인구수 총합을 계산한다
group_df= data.groupby(['광역시도','행정구역'])['인구수'].sum()
# 광역시도와 행정구역별로 집계한 인구수 데이터를 엑셀 파일로 저장하고 시트 이름을 '인구수합계'로 지정한다
group_df.to_excel('data/광역시도별행정구역별인구수합계.xlsx', sheet_name='인구수합계')
```

#### 상관계수 (Correlation Coefficient)
* 인구수와 면적 데이터간에 관련성이 있는지 살펴보기 위해서 상관계수 구하기
* 상관계수 값은 -1 ~ 1 사이의 값이며, 0에 가까울 수록 관련성이 낮고, 1에 가까울 수록 관련성이 높다. 
* 음수는 반비례 (면적이 넓은 반면 인구수는 적은 경우), 양수는 비례 (면적이 넓고, 인구수도 높은 경우)
* corr() 함수
* 상관계수는 두 데이터가 함께 어떻게 변하는지를 숫자로 나타낸 값임.
* 여기서는 인구수와 면적이 서로 관련이 있는지를 보기 위해 사용함.(예: 면적이 넓으면 인구수가 많은지 등)
* 값의 범위는 -1 ~ 1
* 0에 가까우면 → 거의 관련 없음
* 1에 가까우면 → 면적이 넓을수록 인구수도 많음(같이 증가)
* -1에 가까우면 → 면적이 넓을수록 인구수는 적음(반대로 움직임)
* 양수는 비례 관계, 음수는 반비례 관계를 의미함.
* 파이썬에서는 corr() 함수로 쉽게 계산할 수 있음.

#### Groupby 한 DataFrame에 집계함수를 여러개 설정하기
* agg() 함수 , agg() 함수는 그룹으로 묶인 데이터에 여러 집계(통계) 함수를 한 번에 적용하는 함수입니다.
* agg(['max','mean','std'])

```python
# agg() 함수 사용하여 여러개의 통계함수를 사용합니다. 
# fillna(0) NaN 을 0 으로 변경해 줌
# 광역시도별로 인구수를 그룹화한 뒤 각 지역의 최대값, 평균, 표준편차를 한 번에 계산한다
data.groupby('광역시도')['인구수'].agg(['max','mean','std']).fillna(0)
```

```python
# agg() 함수 사용 - 그룹핑한 데이터를 여러개의 집계함수를 사용해서 새로운 DataFrame 객체 생성하기
group_agg_df= data.groupby('광역시도')['인구수'].agg(['max','mean','std']).fillna(0)
group_agg_df.sort_values(by='std',ascending=False)
```

```python
# 광역시도 이름을 인자로 받아서 인구수와 면적을 그려주는 함수
def show_pop_area(sido_name):
    # 입력받은 광역시도에 해당하는 DataFrame 객체생성
    sido_df = data.loc[data['광역시도'] == sido_name]
    # Figure와 Axes 객체 생성
    figure, (axes1,axes2) = plt.subplots(nrows=2, ncols=1)
    # Figure size 확대
    figure.set_size_inches(18,12)
    
    pop_plot = sns.barplot(x='행정구역', y='인구수', data=sido_df.sort_values(by='인구수',ascending=False), ax=axes1)
    pop_plot.set_title(f'{sido_name} 행정구역별 인구수')
    
    area_plot = sns.barplot(x='행정구역', y='면적', data=sido_df.sort_values(by='면적',ascending=False), ax=axes2)
    area_plot.set_title(f'{sido_name} 행정구역별 면적')
    plt.show()    
```

```python
show_pop_area('강원도')
```
