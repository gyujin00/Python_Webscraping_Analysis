## 질문1 :  아래의 url에서 뉴스기사의 링크와 제목을 출력하세요. 

```python
import requests  
from bs4 import BeautifulSoup
# 1. 뉴스 URL 설정
url = 'https://news.daum.net/economy'

# 2. 페이지 요청 및 응답 받기
response = requests.get(url)

# 3. 인코딩 설정 
response.encoding = 'utf-8'

# 4. HTTP 상태 코드 및 응답 내용 확인
print(url)
print(response.status_code)

# 5. BeautifulSoup 객체 생성
soup = BeautifulSoup(response.text, 'html.parser')

# 6. 뉴스 기사 리스트 추출
news_list = soup.select('ul.list_newsheadline2 li')

for li_tag in news_list:
    # a 태그에서 링크 추출
    a_tag = li_tag.find('a')
    if a_tag:
        link = a_tag['href']
        
        # 기사 제목 추출 
        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        if strong_tag:
            title = strong_tag.text.strip() # 공백 제거
            
            # 결과 출력
            print(link)
            print(title)
```

## 질문2:  여러개의 section 중 하나를 선택해서 url에서 뉴스기사의 링크와 제목을 출력하는 코드를 함수로 작성하기 

```python
import requests
from bs4 import BeautifulSoup

# 1. 섹션명과 URL 경로를 매핑한 딕셔너리 생성
section_dict = {
    '기후/환경': 'climate', 
    '사회': 'society', 
    '경제': 'economy', 
    '정치': 'politics', 
    '국제': 'world', 
    '문화': 'culture', 
    '생활': 'life', 
    'IT/과학': 'tech', 
    '인물': 'people'
}

# 2. 뉴스 출력 함수 정의
def print_news(section_name):
    # 딕셔너리에서 한글 섹션명에 맞는 영문 경로 추출
    section_path = section_dict.get(section_name)
    
    # 만약 딕셔너리에 없는 섹션명이 입력될 경우 처리
    if not section_path:
        print(f"'{section_name}' 섹션을 찾을 수 없습니다.")
        return

    # URL 생성
    url = f'https://news.daum.net/{section_path}'
    
    # 페이지 요청 및 인코딩 설정
    response = requests.get(url)
    response.encoding = 'utf-8'
    
    # BeautifulSoup으로 파싱
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # 실행 결과 헤더 출력
    print(f"\n======> {url} {section_name} 뉴스 <======")
    
    # 뉴스 기사 리스트 추출 
    news_items = soup.select('ul.list_newsheadline2 li')
    
    for li_tag in news_items:
        # a 태그에서 링크(href) 추출
        a_tag = li_tag.find('a')
        # 제목이 포함된 strong 태그 추출
        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        
        if a_tag and strong_tag:
            link = a_tag['href']
            title = strong_tag.text.strip()
            
            # 링크와 제목 출력
            print(link)
            print(title)


print_news('경제')

```
